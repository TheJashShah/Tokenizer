### Comparing the outputs of Naive Tokenizer and Inverted Tokenizer

The Inverted method of using a chunk hashmap, chunk frequencies, and merging only affected chunks leads to an exponential decrease in the time taken to train the tokenizer on any text-size, for any number of merges. 

We will now compare the outputs of both the methods, to see if they are exactly the same or not.

```python
import os, json, pandas as pd

def get_data(folder_path):
    
    vocab_data = None
    merge_data = None
    results = None
    
    for file in os.listdir(folder_path):
        
        if "vocab" in file:
            
            with open(os.path.join(folder_path, file), "r", encoding="utf-8") as f:
                vocab_data = json.load(f)
                
        if "merges" in file:
            
            with open(os.path.join(folder_path, file), "r", encoding="utf-8") as f:
                merge_data = f.read()
                
        if "results" in file and file.split(".")[-1] == "csv":
            results = pd.read_csv(os.path.join(folder_path, file))
            
    return vocab_data, merge_data, results
```

```python
import os
import json
import pandas as pd

current = os.getcwd()
root = os.path.join(current, "..")

output_folder = os.path.join(root, "output_files")
naive_folder = os.path.join(output_folder, "trial_tokenizer")
optimized_folder = os.path.join(output_folder, "trial_tokenizer_inverted")

naive_vocab, naive_merges, naive_results = get_data(naive_folder)
inverted_vocab, inverted_merges, inverted_results = get_data(optimized_folder)

if naive_vocab == inverted_vocab:
    print("Vocab Match.")
else:
    print("Vocab Mismatch.")

if naive_merges == inverted_merges:
    print("Merges Match.")
else:
    print("Merges Mismatch.")

if naive_results.equals(inverted_results):
    print("Results Match.")
else:
    print("Results Mismatch.")
```
